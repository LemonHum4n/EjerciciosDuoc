```python
# Pregunta 4

def k(t):
    return 1.85*t

import matplotlib.pyplot as plt
import numpy as np

t = np.arange(0, 3567.6)
plt.plot(t, k(t), label = 'k(t)')
plt.title('Nombre del gráfico')
plt.ylabel('Kilometros')
plt.xlabel('Tiempo')
plt.legend()
plt.grid()
plt.show()
```
