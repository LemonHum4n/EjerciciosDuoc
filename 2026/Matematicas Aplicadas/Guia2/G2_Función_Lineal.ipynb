# Problema 1

Un equipo de desarrollo está evaluando la eficiencia de un nuevo algoritmo implementado en *Python*. Han medido el tiempo de ejecución del algoritmo en función de la cantidad de elementos de entrada, obteniendo los resultados que se muestran en la siguiente tabla:

<p><center>

| Cantidad de elementos (unidades) | Tiempo de ejecución (ms) |
|----------------------------------|--------------------------|
|$$ 100 $$                         | $$2 $$                   |
|$$ 200 $$                         | $$4 $$                   |
|$$ 500 $$                         | $$10 $$                  |
|$$ 1.000 $$                        | $$20 $$                  |
|$$ 2.000 $$                        | $$40 $$                  |

</center></p>

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Encuentra la función que relaciona el tamaño de la entrada con el tiempo de ejecución del algoritmo, para esto utiliza la función *polyfit* de *Numpy*.
3. Calcula el tiempo de ejecución para una entrada de $1.500$ elementos.
4. Utiliza la función *fsolve* de *Scipy* para determinar el tamaño de entrada que haría que el tiempo de ejecución sea de $50$ milisegundos.


## 1
La variable dependiente es el tiempo de ejecucion en ms, la variable independiente es la cantidad de elementos de entrada en Unidades

## 2

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

x = np.array([100,200,500,1000,2000])
y = np.array([2,4,10,20,40])
#Si la función es de grado 1
pendiente, intercepto = np.polyfit(x, y, 1)

print(f"{pendiente:.2f} {intercepto:.2f}")
```

## 3

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

x = np.array([100,200,500,1000,2000])
y = np.array([2,4,10,20,40])
#Si la función es de grado 1
pendiente, intercepto = np.polyfit(x, y, 1)

def T(n):
    return pendiente*n + intercepto

print(f"El tiempo de ejecucion para una entrada de 1500 elementos seria de {T(1500):.0f}ms")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

x = np.array([100,200,500,1000,2000])
y = np.array([2,4,10,20,40])
#Si la función es de grado 1
pendiente, intercepto = np.polyfit(x, y, 1)

def T(n):
    return pendiente*n + intercepto - 50


n = 0
solucion = fsolve(T, n)

print(f"el tamaño de entrada a los 50 milisegundos seria de {solucion[0]:.0f} unidades.")

```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

x = np.array([100,200,500,1000,2000])
y = np.array([2,4,10,20,40])
#Si la función es de grado 1
pendiente, intercepto = np.polyfit(x, y, 1)

def T(n):
    return pendiente*n + intercepto

n = np.arange(0, 2000.01, 100)
plt.plot(n, T(n), label = 'T(n)')
plt.title('Nombre del gráfico')
plt.ylabel('Tiempo de ejecucion en milisegundos')
plt.xlabel('Tamaño de la entrada en unidades')
plt.legend()
plt.grid()
plt.show()
```

# Problema 2

Un equipo de TI está monitoreando el tráfico de datos en la red de la empresa. Han registrado el volumen de los datos transferidos y el tiempo que tomó cada transferencia en la siguiente tabla de datos:

<p><center>

|Datos transferidos (GB) | Tiempo de transferencia (min)|
|------------------------|------------------------------|
|$$5$$                   |$$10$$                        |
|$$10$$                  |$$20$$                        |
|$$25$$                  |$$50$$                        |
|$$50$$                  |$$100$$                       |
|$$100$$                 |$$200$$                       |

</center></p>

1. Defina variable dependiente e independiente, indicando unidad de medida.
2. Utilizando la función *polyfit* de *Numpy* encuentra la función que modela el tiempo de transferencia $T(x)$ en función del volumen de los datos transferidos.
3. Interprete la pendiente en el contexto del problema.
4. ¿Cuánto tiempo toma una transferencia de $73,2$ GB?
5. Utiliza la función *fsolve* de *Scipy* para determinar la cantidad de datos que se puedan transferir en $123,5$ minutos.
6. Grafica la función $T(x)$ indicando el nombre de los ejes y su unidad de medida. Interpreta lo observado. *Para graficar utiliza la biblioteca Matplotlib.*

## 1
La variable dependiente es el tiempo de transferencia en minutos, y la variable independiente son los datos transferidos en GB.

## 2

```python
x = np.array([5,10,25,50,100]) #independiente
y = np.array([10,20,50,100,200]) #dependiente
#Si la función es de grado 1
pendiente, intercepto = np.polyfit(x, y, 1)

print(f"pendiente={pendiente:.0f} intercepto={intercepto:.0f}")
```

## 3 
La pendiente señala que por cada GB transferido el tiempo de transferencia aumenta en dos minutos.

## 4

```python
def T(d):
    return pendiente*d + intercepto

print(f"El tiempo para hacer una tranferencia de 73,2GB es de {T(73.2):.1f} minutos")
```

## 5

```python
def T(d):
    return pendiente*d + intercepto - 123.5

#Valor inicial de la aproximación:
d = 0
solucion = fsolve(T, d)

print(f"La cantidad de datos que se transfieren en 123.5 minutos son en total {solucion[0]:.1f} GB")
```

## 6

```python
def T(d):
    return pendiente*d + intercepto

d = np.arange(0, 100.01)
plt.plot(d, T(d), label = 'T(d)')
plt.title('Nombre del gráfico')
plt.ylabel('Tiempo en minutos')
plt.xlabel('Datos transferidos en GB')
plt.legend()
plt.grid()
plt.show()
```

La relacion entre ambas variables es lineal y positiva, cuanto mayor sea el volumen de datos, mayor es el tiempo de transferencia.

# Problema 3


Una empresa de tecnología quiere optimizar el tiempo de carga y descarga de archivos en su servicio de almacenamiento en la nube. Los tiempos medios para diferentes tamaños de archivos son los siguientes:

<p><center>

| Tamaño del archivo (MB) | Tiempo de carga (s) | Tiempo de descarga (s)|
|-------------------------|---------------------|-----------------------|
|$$50$$                   |$$10$$               |$$8$$                  |
|$$100$$                  |$$20$$               |$$16$$                 |
|$$250$$                  |$$50$$               |$$40$$                 |
|$$500$$                  |$$100$$              |$$80$$                 |
|$$1.000$$                 |$$200$$              |$$160$$                |

</center></p>

1. Encuentra las funciones que modelan el tiempo de carga $f(x)$ y el tiempo de descarga $g(x)$ en función del tamaño del archivo $x$. Utiliza la función *polyfit* de *Numpy* (debes encontrar dos funciones).
2. Define variable dependiente e independiente, para ambas funciones, indicando su unidad de medida.
3. Calcula el tiempo de carga y descarga para un archivo de $750$ MB.
4. Utiliza la función *fsolve* de *Scipy* para determinar el tamaño del archivo si el tiempo de carga es de $163$ segundos.
5. Un trabajador insiste en que si el tiempo de descarga es de $195$ segundos, el tamaño del archivo es $1.200$ MB. ¿Tiene razón? Utiliza *Python* para determinar.
6. Grafica ambas funciones en el mismo gráfico, indicando nombre de los ejes y unidades de medida. Utiliza la librería *Matplotlib* para graficar.

## 1

```python
# tiempo de carga

a = np.array([50,100,250,500,1000]) #independiente - archivo
f = np.array([10,20,50,100,200]) #dependiente - carga
g = np.array([8,16,40,80,160]) #dependiente - descarga
#Si la función es de grado 1
pendienteCarga, interceptoCarga = np.polyfit(a, f, 1)
pendienteDescarga, interceptoDescarga = np.polyfit(a, g, 1)

print(f"Pendiente: {pendienteCarga:.2f} y {pendienteDescarga:.2f} Intercepto Carga: {interceptoCarga:.2f} Intercepto Descarga: {interceptoDescarga:.2f}")

```

## 2

Las variables dependientes son el tiempo de Carga y el tiempo de Descarga en segundos. Y la variable independiente es el tamaño del archivo en MB.

## 3

```python
def f(a):
    return pendienteCarga*a + interceptoCarga

def g(a):
    return pendienteDescarga*a + interceptoDescarga

print(f"El tiempo de carga del archivo 750MB es de {f(750):.2f} segundos, y el tiempo de descarga es de {g(750):.2f} segundos")
```

## 4

```python
def f(a):
    return pendienteCarga*a + interceptoCarga - 163

#Valor inicial de la aproximación:
a = 0
solucion = fsolve(f, a)

print(f"El tamaño del archivo con una carga de 163 segundos es de {solucion[0]:.2f} MB")
```

## 5

```python
def g(a):
    return pendienteDescarga*a + interceptoDescarga - 195

a = 0
solucion = fsolve(g, a)

print(f"El tamaño del archivo con una descarga de 195 segundos es de {solucion[0]:.2f} MB")
```

Lo que dice el trabajador esta mal, porque el tamaño del archivo tiene que ser de 1218.75 MB para que su tiempo de descarga sea de 195 segundos.

## 6

```python
def f(a):
    return pendienteCarga*a + interceptoCarga

def g(a):
    return pendienteDescarga*a + interceptoDescarga

a = np.arange(0, 1000)
plt.plot(a, f(a), label = 'f(x) Tiempo Carga')
plt.plot(a, g(a), label = 'g(x) Tiempo Descarga')
plt.title('Nombre del gráfico')
plt.ylabel('Tiempo en segundos (s)')
plt.xlabel('Tamaño del archivo (MB)')
plt.legend()
plt.grid()
plt.show()
```
